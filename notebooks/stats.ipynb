```python
# Parâmetro: pasta de resultado a ser analisada
RESULT_DIR = "result_0004"
```

```python
import os
import pandas as pd
from pathlib import Path

NOTEBOOK_DIR = Path(os.getcwd())
PROJECT_ROOT = NOTEBOOK_DIR.parent
RESULTS_PATH = PROJECT_ROOT / 'project' / 'results' / RESULT_DIR

print(f'Resultados: {RESULTS_PATH}')
```

```python
registros = []
for dataset_dir in sorted(RESULTS_PATH.iterdir()):
    if not dataset_dir.is_dir() or dataset_dir.name == 'temp':
        continue
    dataset = dataset_dir.name
    for csv_file in sorted(dataset_dir.glob('*.csv')):
        df = pd.read_csv(csv_file)
        df['dataset'] = dataset
        df['instance'] = csv_file.stem
        registros.append(df)

df_raw = pd.concat(registros, ignore_index=True)
print(f'Total de linhas carregadas: {len(df_raw)}')
```

```python
# Filtra apenas execuções viáveis
df_ok = df_raw[df_raw['status'] == 'feasible'].copy()
print(f'Execuções com status feasible: {len(df_ok)} de {len(df_raw)}')

# Tabela com as colunas solicitadas
df_table = df_ok[['dataset', 'instance', 'algo_id', 'seed', 'objective']].copy()
df_table.columns = ['dataset', 'instance', 'algorithm', 'seed', 'objective']

display(df_table.reset_index(drop=True))
```
